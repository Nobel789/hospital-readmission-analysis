Step 2: Create Toy Healthcare Data

We simulate monthly admissions for 3 departments.

```python
import pandas as pd

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "ICU": [30, 40, 55, 70, 90, 120, 150, 180, 220, 260, 300, 350],
    "General Ward": [50, 60, 75, 90, 110, 140, 170, 200, 240, 280, 320, 380],
    "Emergency": [20, 30, 45, 60, 80, 100, 130, 160, 200, 240, 280, 320]
}

df = pd.DataFrame(data)
print(df)
```

Create Area Chart

```python
import matplotlib.pyplot as plt

plt.figure()

plt.stackplot(
    df["Month"],
    df["ICU"],
    df["General Ward"],
    df["Emergency"],
    labels=["ICU", "General Ward", "Emergency"]
)

plt.title("Healthcare Analytics: Patient Admissions")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.legend()

plt.show()
```
